### 3. Generating a sinusoid

3.1. Create a sinusoid of the same duration than the reference signal and same frequency than the fundamental frequency. Plot it along with the audio signal of the reference recording. Plot the same time segment that 2.3. Note that the sampling rate should be the same for both signals. Try to find the values of Amplitude and initial Phase that make the two signal segments plotted to match as close as possible (do not attempt to imitate the whole signal). Matching the phase is a bit tricky.



```python
import numpy as np

# 1. Calculate total duration from the original audio array
total_duration = len(audio_signal) / sample_rate

# 2. Create the time array for the full audio duration
t = np.linspace(0, total_duration, len(audio_signal), endpoint=False)

# 3. Define sinusoid parameters
# Adjust 'amplitude' and 'phase' below to fit your specific plot
amplitude = 0.5        # Match the peak amplitude of the sustain phase
f_0 = 293.68           # Fundamental frequency measured in Section 2.1 (Hz)
phase = np.pi / 4      # Initial phase offset in radians (e.g., pi/4, pi/2, etc.)

# 4. Synthesize the pure sinusoid
sinusoid_signal = amplitude * np.sin(2 * np.pi * f_0 * t + phase)

# 5. Define the 3-period time segment (same as in 2.3)
t_start = 0.15 
num_periods = 3
T = 1 / f_0
t_end = t_start + (num_periods * T)

# 6. Plot both signals together over the selected window
plot_signals(
    [audio_signal, sinusoid_signal], 
    sample_rate, 
    t_start=t_start, 
    t_end=t_end, 
    title="Comparison: Original Kalimba Signal vs. Synthesized Fundamental Sinusoid"
)
```
